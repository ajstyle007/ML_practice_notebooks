```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier
```

```python
X, y = make_classification(n_samples=1000, n_features=10, n_informative=3)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
X_train.shape, X_test.shape
```

```python
# Lets train a simple  decision tree
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

# Bagging 

```python
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.25,
    bootstrap=True,
    random_state=23
)
```

```python
bag.fit(X_train, y_train)
```

```python
y_pred = bag.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```

# Bagging using SVM

```python
bag1 = BaggingClassifier(
    estimator=SVC(),
    n_estimators=100,
    max_samples=0.25,
    bootstrap=True,
    random_state=23
)
```

```python
bag1.fit(X_train, y_train)
```

```python
y_pred1 = bag1.predict(X_test)
accuracy_score(y_test, y_pred1)
```

## Pasting

```python
pst = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.25,
    bootstrap=False,
    random_state=23
)
```

```python
pst.fit(X_train, y_train)
```

```python
y_pred = pst.predict(X_test)
accuracy_score(y_test, y_pred)
```

## Random Subspaces

```python
rsub = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=100,
    max_samples=0.25,
    bootstrap=False,
    random_state=23,
    bootstrap_features=True,
    max_features=0.5
)
```

```python
rsub.fit(X_train, y_train)
```

```python
y_pred = rsub.predict(X_test)
```

```python
accuracy_score(y_test, y_pred)
```
