# OOB (out of Bag) [score] evaluation

### What is it?
### The OOB score is an estimate of the model's accuracy on unseen data.
### It leverages the data left out of the bootstrap sampling process used in creating each decision tree within the Random Forest.

```python
# On average, about 63% of the original data points are included in each bootstrap sample.
# The remaining 37% of the data points are not included in the bootstrap sample for that tree 
# and are referred to as the "out-of-bag" data.
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
X = df.iloc[:, 0:13]
y = df.iloc[:,13]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
rf = RandomForestClassifier(oob_score=True)
```

```python
rf.fit(X_train, y)
```
