# Difference Between Random forest and Bagging classifier

```python
# 1. first difference is in bagging we can use different algorithms as base estimators but in random forest we only use DT
# 2. Second and main difference is when we do column sampling in both then bagging use only selected features for node splliting in decision tree
# but in random forest there is random sampling om each node for introducing more diversity among trees.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import plot_tree
```

```python
X, y = make_classification(n_features=5, n_samples=100, n_informative=2, random_state=23)
```

```python
df = pd.DataFrame(X, columns=["col1", "col2", "col3", "col4", "col5"])
df["target"] = y
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:5], df.iloc[:,-1], test_size=0.2, random_state=23)
```

```python
X_train.shape, X_test.shape
```

```python
# First we will train the model with Bagging classifier with estimator Decision tree
```

```python
bag = BaggingClassifier(max_features=2)
bag.fit(X_train, y_train)
```

```python
y_pred = bag.predict(X_test)
```

```python
plt.figure(figsize=(12,8))
plot_tree(bag.estimators_[0])
plt.show()
```

```python
bag.estimators_[0]
```

```python
rf = RandomForestClassifier(max_features=2)
rf.fit(X_train, y_train)
```

```python
plt.figure(figsize=(12,8))
plot_tree(rf.estimators_[0])
plt.show()
```
