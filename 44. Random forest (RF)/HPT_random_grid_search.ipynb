```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, precision_score, confusion_matrix
```

```python
df = pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
X = df.iloc[:,0:13]
y = df.iloc[:,13]
```

```python
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
X_train.shape, X_test.shape
```

```python
rf = RandomForestClassifier()
gb = GradientBoostingClassifier()
lor = LogisticRegression()
svm = SVC()
```

```python
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
gb.fit(X_train, y_train)
y_pred = gb.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
lor.fit(X_train, y_train)
y_pred = lor.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
svm.fit(X_train, y_train)
y_pred = svm.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
#Now we will check it by using cross validation
```

```python
np.mean(cross_val_score(rf, X, y, cv=10, scoring="accuracy"))
```

```python
np.mean(cross_val_score(lor, X, y, cv=10, scoring="accuracy"))
```

```python
rf = RandomForestClassifier(max_samples=0.75, bootstrap=True, random_state=23)
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
np.mean(cross_val_score(rf, X, y, cv=10, scoring="accuracy"))
```

# Hyperparameter Tuning

## GridSearchCV

```python
# Number of decision trees
n_estimators = [20, 50, 100, 150]

# number of features to be consider at evry split
max_features = 0.2, [0.5, 0.75, 0.85, 1]

# number of samples 
max_samples = [0.5, 0.75, 1.0]

# depth of trees 
max_depth = [2, 8, None]
```

```python
params = {
    "n_estimators" : n_estimators,
    "max_features" : max_features,
    "max_samples" : max_samples,
    "max_depth" : max_depth,
}
print(params)
```

```python
rf = RandomForestClassifier()
```

```python
rf_grid = GridSearchCV(estimator=rf, param_grid=params, cv=5, verbose=2, n_jobs=-1)
```

```python
rf_grid.fit(X_train, y_train)
```

```python
rf_grid.best_params_
```

```python
rf_grid.best_score_
```

# RandomSearchCV

```python
# Number of decision trees
n_estimators = [20, 50, 100, 150]

# number of features to be consider at evry split
max_features = 0.2, [0.5, 0.75, 0.85, 1]

# number of samples 
max_samples = [0.5, 0.75, 1.0]

# depth of trees 
max_depth = [2, 8, None]

# bootstrap samples
bootstrap = [True, False]

# minimum samples required to split a node
min_samples_split = [2, 5]

# min samples reuired at each leaf node
min_samples_leaf = [1, 2]
```

```python
params = {
    "n_estimators" : n_estimators,
    "max_features" : max_features,
    "max_samples" : max_samples,
    "max_depth" : max_depth,
    "bootstrap" : bootstrap,
    "min_samples_split" : min_samples_split,
    "min_samples_leaf" : min_samples_leaf
}
print(params)
```

```python
rf = RandomForestClassifier()
rf_grid = GridSearchCV(estimator=rf, param_grid=params, cv=5, verbose=2, n_jobs=-1)
```

```python
rf_grid.fit(X_train, y_train)
```

```python
rf_grid.best_params_
```

```python
rf_grid.best_score_
```
