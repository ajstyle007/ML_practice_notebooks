```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("train.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"], inplace=True)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop("Survived", axis =1), df["Survived"], test_size = 0.2, random_state=0)
```

```python
print(X_train.shape)
X_train.head(3)
```

```python
X_test.shape
```

```python
X_train["Embarked"].value_counts()
```

```python
print(y_train.shape)
y_train.head()
```

```python
df.isnull().sum()
```

```python
# applying imputation on missing values column
si_age = SimpleImputer()
si_emb = SimpleImputer(strategy="most_frequent")

X_train_age = si_age.fit_transform(X_train[["Age"]])
X_train_emb = si_emb.fit_transform(X_train[["Embarked"]])

X_test_age = si_age.transform(X_test[["Age"]])
X_test_emb = si_emb.transform(X_test[["Embarked"]])
```

```python
X_train_age
```

```python
# applying onehotencoding on nominal categorical columns
ohe_sex = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
ohe_emb = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

X_train_sex = ohe_sex.fit_transform(X_train[["Sex"]])
X_train_embarked = ohe_emb.fit_transform(X_train_emb)

X_test_sex = ohe_sex.transform(X_test[["Sex"]])
X_test_embarked = ohe_emb.transform(X_test_emb)
```

```python
print(X_train_sex.shape)
X_train_sex
```

```python
X_train_embarked
```

```python
X_train.head(2)
```

```python
# Now we will concatinate the columns
X_train_rem = X_train.drop(columns=["Sex", "Age", "Embarked"])
X_test_rem = X_test.drop(columns=["Sex", "Age", "Embarked"])
```

```python
X_train_transformed = np.concatenate((X_train_rem,X_train_sex, X_train_age, X_train_embarked), axis = 1)
X_test_transformed = np.concatenate((X_test_rem,X_test_sex, X_test_age, X_test_embarked), axis = 1)
```

```python
X_train_transformed.shape
```

```python
X_test_transformed.shape
```

```python
clf = DecisionTreeClassifier()
clf.fit(X_train_transformed, y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```
