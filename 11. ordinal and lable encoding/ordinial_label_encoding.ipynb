```python
import pandas as pd
```

```python
df = pd.read_csv("customer.csv")
```

```python
df.head()
```

```python
# In Encoding categorical data 
# We have two types of data
# 1. Numerical data
# 2. Categorical data

# And we have two types of categorical data
# 1. Nominal data (With this data we use One Hot Encodingg to encode this data.)
# 2. Ordinal data (With this data we use Ordinal encoding with the features and label encoding with the label(target) column.)

# in above data 
# age > numerical column
# gender > Nominal Categorical column
# review > Ordinal "" column
# education > Ordinal "" column
# purchased > Nominal "" column
```

```python
df1 = df.iloc[:, 2:]
```

```python
df1.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df1.drop("purchased", axis = 1), df1["purchased"], test_size = 0.2, random_state=0)
```

```python
X_train.head()
```

# Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories = [["Poor", "Average", "Good"], ["School", "UG", "PG"]])
```

```python
oe.fit(X_train)
```

```python
X_train_tran = oe.transform(X_train)
X_test_tran = oe.transform(X_test)
```

```python
X_train_tran
```

```python
oe.categories_
```

# Label Encoding 

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
le.fit(y_train)
```

```python
y_train_tran = le.transform(y_train)
y_test_tran = le.transform(y_test)
```
