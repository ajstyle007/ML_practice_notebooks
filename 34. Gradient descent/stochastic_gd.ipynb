```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
import time
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)
```

```python
X_train.shape
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
class SGDreg:

    def __init__(self, learning_rate, epochs):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):
        # initializing the coeffs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        #updating all the coeffs and intercepts
        for i in range(self.epochs):
            for j in range(X_train.shape[0]):

                index = np.random.randint(0, X_train.shape[0])
                y_hat = np.dot(X_train[index], self.coef_) + self.intercept_
                intercept_slope = -2 * (y_train[index] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_slope)

                coef_slope = -2 * np.dot((y_train[index] - y_hat) , X_train[index]) 
                self.coef_ = self.coef_ - (self.lr * coef_slope)
                
        print(self.intercept_, self.coef_)      

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
sgd = SGDreg(learning_rate=0.01, epochs=60)
```

```python
start = time.time()
sgd.fit(X_train, y_train)
print("time taken", time.time() - start)
```

```python
y_pred = sgd.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```
