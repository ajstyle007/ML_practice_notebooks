```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("placement.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
plt.scatter(df["cgpa"], df["package"])
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("Scatter Plot.")
plt.show()
```

```python
X = df.iloc[:,0:1]
y = df.iloc[:,1]
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr
```

```python
X_test.head()
```

```python
y_test.head()
```

```python
lr.predict(X_test.iloc[0].values.reshape(1,1))
```

```python
plt.scatter(df["cgpa"], df["package"])
plt.plot(X_train, lr.predict(X_train), c="red", lw= 3) # lw = linewidth
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("Scatter Plot.")
plt.show()
```

```python
# y = mx + b
# m >> slope
# b >> intercept
```

```python
m = lr.coef_
m
```
