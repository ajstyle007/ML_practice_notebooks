```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.impute import MissingIndicator, SimpleImputer
```

```python
df = pd.read_csv("train.csv", usecols=["Age", "Fare", "Survived"])
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().mean()*100
```

```python
X = df.drop("Survived", axis = 1)
y = df["Survived"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=43)
```

```python
X_train.head()
```

```python
si = SimpleImputer()
X_train_trf  = si.fit_transform(X_train)
X_test_trf = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()
```

```python
clf.fit(X_train_trf, y_train)
```

```python
y_pred = clf.predict(X_test_trf)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
mi = MissingIndicator()
```

```python
mi.fit(X_train)
```

```python
X_train_miss = mi.fit_transform(X_train)
X_test_miss = mi.transform(X_test)
```

```python
mi.features_
```

```python
X_train["Age_NA"] = X_train_miss
```

```python
X_test["Age_NA"] = X_test_miss
```

```python
X_train.head()
```

```python
si = SimpleImputer()
```

```python
X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
clf = LogisticRegression()
```

```python
clf.fit(X_train_trf2, y_train)
```

```python
y_pred = clf.predict(X_test_trf2)
```

```python
accuracy_score(y_test, y_pred)
```

```python
si = SimpleImputer(add_indicator=True)
```

```python
X_train_trf3 = si.fit_transform(X_train)
X_test_trf3 = si.transform(X_test)
```

```python
clf = LogisticRegression()
```

```python
clf.fit(X_train_trf3, y_train)
```

```python
y_pred = clf.predict(X_test_trf3)
```

```python
accuracy_score(y_test, y_pred)
```
