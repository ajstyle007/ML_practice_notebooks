```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("train.csv")
df.head()
```

# 1. Categorical data

## a. countplot

```python
sns.countplot(df, x = df["Survived"])
df["Survived"].value_counts()
```

```python
#palette = ["#F72585", "#7209B7", "#3A0CA3"]
#sns.countplot(df, x = df["Pclass"], palette = "Set1")
sns.countplot(df, x = df["Embarked"], palette = "Set2")
#df["Survived"].value_counts().plot(kind = "bar")
plt.show()
```

## b. Pie chart 

```python
# if we want to see the same information in terms of percentage.
#df["Survived"].value_counts().plot(kind = "pie", autopct="%.2f")
#df["Pclass"].value_counts().plot(kind = "pie", autopct="%.2f")
#df["Embarked"].value_counts().plot(kind = "pie", autopct="%.2f")
df["Sex"].value_counts().plot(kind = "pie", autopct="%.2f")
```

# 2. Numerical data

## a. Histogram

```python
# histogram provides the distribution of the data
plt.hist(df["Age"])
#plt.hist(df["Age"], bins=20)
#plt.hist(df["Fare"], bins = 5)
```

## b. Distplot

```python
# distplot provides the probability density of the data
sns.distplot(df, x = df["Age"])
#sns.distplot(df, x = df["Fare"])
```

## c. boxplot

```python
# it gives five number summary of the data and helps to find the outliers in the data
#sns.boxplot(df, x = df["Age"])
sns.boxplot(df, x = df["Fare"])
```
