```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import recall_score, precision_score, f1_score, confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
hd = pd.read_csv("heart.csv")
```

```python
hd.head()
```

```python
X = hd.iloc[:,0:-1]
```

```python
X.head()
```

```python
y = hd.iloc[:,-1]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()
```

```python
clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
```

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
```

```python
X_train.shape, X_test.shape
```

```python
print("Logistic Regression Model: ")
print("-"*50)
print(pd.DataFrame(confusion_matrix(y_test, y_pred1), columns=list(range(0,2))))
print("Precision score: ", precision_score(y_test, y_pred1))
print("recall score: ", recall_score(y_test, y_pred1))
print("f1 score: ", f1_score(y_test, y_pred1))
```

```python
print("Decision Tree Model: ")
print("-"*50)
print(pd.DataFrame(confusion_matrix(y_test, y_pred2), columns=list(range(0,2))))
print("Precision score: ", precision_score(y_test, y_pred2))
print("recall score: ", recall_score(y_test, y_pred2))
print("f1 score: ", f1_score(y_test, y_pred2))
```

```python
l_df = pd.DataFrame(confusion_matrix(y_test, y_pred1), columns=list(range(0,2)))
```

```python
print("precsion: ")
print(l_df[0][0])
print(l_df[0][1])
print(l_df[0][0] / (l_df[0][0] + l_df[0][1]))
```

# For Multiclass classification

```python
mnist = pd.read_csv("mnist.csv")
```

```python
mnist.head()
```

```python
X = mnist.iloc[:, 1:]
X.head()
```

```python
y = mnist.iloc[:,0]
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
lr = LogisticRegression()
dt = DecisionTreeClassifier()
```

```python
lr.fit(X_train, y_train)
dt.fit(X_train, y_train)
```

```python
y_pred3 = lr.predict(X_test)
y_pred4 = dt.predict(X_test)
```

```python
print("Confusion Matrix for Logistic Regression: ")
print("-"*50)
pd.DataFrame(confusion_matrix(y_test, y_pred3))
```

```python
print(precision_score(y_test, y_pred3, average=None))
print(precision_score(y_test, y_pred3, average="macro"))
print(precision_score(y_test, y_pred3, average="weighted"))
```

```python
print(recall_score(y_test, y_pred3, average=None))
print(recall_score(y_test, y_pred3, average="macro"))
print(recall_score(y_test, y_pred3, average="weighted"))
```

```python
print(f1_score(y_test, y_pred3, average=None))
print(f1_score(y_test, y_pred3, average="macro"))
print(f1_score(y_test, y_pred3, average="weighted"))
```

```python
print("Confusion Matrix for DT: ")
print("-"*50)
pd.DataFrame(confusion_matrix(y_test, y_pred4))
```

```python
print("Precision score for DT: ")
print("-"*50)
print(precision_score(y_test, y_pred4, average=None))
print(precision_score(y_test, y_pred4, average="macro"))
print(precision_score(y_test, y_pred4, average="weighted"))
print("-"*50)

print("Recall score for DT: ")
print("-"*50)
print(recall_score(y_test, y_pred4, average=None))
print(recall_score(y_test, y_pred4, average="macro"))
print(recall_score(y_test, y_pred4, average="weighted"))
print("-"*50)

print("f1 score for DT: ")
print("-"*50)
print(f1_score(y_test, y_pred4, average=None))
print(f1_score(y_test, y_pred4, average="macro"))
print(f1_score(y_test, y_pred4, average="weighted"))
print("-"*50)
```

```python
from sklearn.metrics import classification_report
```

```python
print("Classification Report for Logistic Regression: ")
print("-"*53)
print(classification_report(y_test, y_pred3))
```

```python
print("Classification Report for DT: ")
print("-"*53)
print(classification_report(y_test, y_pred4))
```
