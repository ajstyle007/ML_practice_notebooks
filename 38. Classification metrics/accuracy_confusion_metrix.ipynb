```python
import pandas as pd
data = pd.read_csv('heart.csv')
```

```python
data.sample(5)
```

```python
data.shape
```

```python
X = data.iloc[:,0:-2]
X.head()
```

```python
y = data.iloc[:,-1]
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=23)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()
```

```python
clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
```

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Logisgtic Accuracy: ", 100* accuracy_score(y_test, y_pred1))
print("Decision tree Accuracy: ", 100* accuracy_score(y_test, y_pred2))
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print("Logistic confusion matrix:\n", confusion_matrix(y_test, y_pred1))
print("Decision tree confusion matrix:\n", confusion_matrix(y_test, y_pred1))
```

```python
y.shape
```

```python
print("Logisgtic Accuracy: ", 100* accuracy_score(y_test, y_pred3))
print("Decision tree Accuracy: ", 100* accuracy_score(y_test, y_pred4))
```
