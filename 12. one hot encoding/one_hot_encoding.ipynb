# OneHotEncoding with Top(large number of ) Categories

```python
counts = df["brand"].value_counts()
```

```python
df["brand"].nunique()
```

```python
thresold = 100
```

```python
repl = counts[counts <= thresold].index
```

```python
repl
```

```python
pd.get_dummies(df["brand"].replace(repl, "uncommon"), dtype=np.int32)
```

```python
pd.get_dummies(df["brand"].replace(repl, "uncommon"), dtype=np.int32).sample(5)
```
