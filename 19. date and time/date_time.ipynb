# 1. Extract year

```python
date["year"] = date["date"].dt.year
```

# 2. Extract month

```python
date["month"] = date["date"].dt.month
```

```python
date.head()
```

```python
date["month_name"] = date["date"].dt.month_name()
```

```python
date.sample(4)
```

# 3. Extract day

```python
date["day"] = date["date"].dt.day
```

```python
date.head()
```

```python
# day of week
date["day_dow"] = date["date"].dt.dayofweek
```

```python
date.sample(5)
```

```python
date.drop(columns=["day_name"], inplace=True)
```

```python
date.sample(3)
```

```python
# day of week name
date["day_name"] =date["date"].dt.day_name()
```

```python
date.sample(4)
```

```python
# day is weekend or not?
date["weekend"] = np.where(date["day_name"].isin(["Saturday", "Sunday"]),1, 0)
```

```python
date.head()
```

```python
# Extracting week of the year (there are 52 or 53 weeks in a year)
```

```python
date["date_week"] = date["date"].dt.isocalendar().week
```

```python
date.head()
```

```python
# Extracting the quarter of the year
```

```python
date["quarter"] = date["date"].dt.quarter
```

```python
date.head()
```

```python
# Extract Semester
```

```python
date["semester"] = np.where(date["quarter"].isin([1,2]), 1, 2)
```

```python
date.sample(3)
```

```python
import pandas as pd

# Create a DataFrame with a datetime column
data = {'timestamp': ['2024-02-01', '2024-02-10', '2024-03-15']}
df = pd.DataFrame(data)

# Convert the 'timestamp' column to datetime format
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Access the week number using the dt accessor
df['week_number'] = df['timestamp'].dt.isocalendar().week

# Print the DataFrame with the week number
print(df)
```
