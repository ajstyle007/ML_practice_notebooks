## Bernoulli Distribution

- Bernoulli Distribution is a type of discrete probability distribution where every experiment conducted asks a question that can be answered only in yes or no. In other words, the random variable can be 1 with a probability p or it can be 0 with a probability (1 - p). Such an experiment is called a Bernoulli trial. A pass or fail exam can be modeled by a Bernoulli Distribution.

- there is only one trial in bernoulli distribution

### Probability Mass Function for Bernoulli Distribution

- We calculate the probability mass function for a Bernoulli distribution. The probability that a discrete random variable will be exactly equal to some value is given by the probability mass function. The formula for pmf, f, associated with a Bernoulli random variable over possible outcomes 'x' is given as follows:

- PMF = f(x, p) = p if x = 1 
- PMF = f(x, p) = q = 1- p if x = 0 

- We can also express this formula as,

- f(x, p) = (p^x) * (1 - p)^1 - x
- x ϵ {0, 1}

# Binomial Distribution

- The binomial distribution represents the probability for 'x' successes of an experiment in 'n' trials, given a success probability 'p' for each trial at the experiment.

### criteria
- the process consist of n trials
- only two exclusive outcomes are possible, a success and a failure
- P(success) = p and P(failure) = 1-p , and it is fixed from trial to trial
- the trials are independent

- binomial distribution consist of "n" numbers trials of bernoulli distribution

- The binomial distribution formula is for any random variable X, given by;  
- P(x:n,p) = (nCx) * p^x (1-p)^(n-x) Or 
- P(x:n,p) = (nCx) * p^x (q)(n-x) where (q = 1 - p)

- where,
- n = the number of experiments
- x = 0, 1, 2, 3, 4, …
- p = Probability of success in a single experiment
- q = Probability of failure in a single experiment (= 1 – p)

- In a single experiment when n = 1, the binomial distribution is called a Bernoulli distribution.

# Graph of PDF 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 10 # number of trials 
p = 0.5 # probabilty of success
size = 1000 # number of samples to generate
```

```python
binomial_dist = np.random.binomial(n, p, size)
```

```python
plt.hist(binomial_dist, density=True)
plt.show()
```

```python
p1 = 0.5
p2 = 0.9
p3 = 0.1

binomial_dist1 = np.random.binomial(n, p1, size)
binomial_dist2 = np.random.binomial(n, p2, size)
binomial_dist3 = np.random.binomial(n, p3, size)

color = ["red", "green", "orange"]
plt.hist(binomial_dist1, density=True, bins=50, color=color[0])
plt.hist(binomial_dist2, density=True, bins=50, color=color[1])
plt.hist(binomial_dist3, density=True, bins=50, color=color[2])
plt.legend(color, loc="upper center")
plt.show()
```
