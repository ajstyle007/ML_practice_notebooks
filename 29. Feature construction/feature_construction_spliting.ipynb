```python
import pandas as pd
import numpy as np
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv("train.csv", usecols=["Survived","Age", "Pclass", "SibSp", "Parch"])
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
X = df.iloc[:, 1:]
y = df.iloc[:, 0]
```

```python
X.head()
```

```python
cross_val_score(LogisticRegression(), X, y, scoring="accuracy", cv=20).mean()
```

# Applying feature construction

```python
X["Family_size"] = X["SibSp"] + X["Parch"] + 1
```

```python
X.head()
```

```python
def myfam(num):
    if num == 1: # alone
        return 0
    elif num > 1 and num <= 4: # small family
        return 1
    elif num > 4: # large family
        return 2
```

```python
myfam(5)
```

```python
X["Family_type"] = X["Family_size"].apply(myfam)
```

```python
X.sample(5)
```

```python
X.drop(columns=["SibSp", "Parch", "Family_size"], inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring="accuracy", cv=20))
```
